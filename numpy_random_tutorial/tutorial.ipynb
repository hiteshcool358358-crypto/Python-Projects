# Random Intro

### Generating random integer and float number

```python
import numpy as np
from numpy import random

# Generates a random integer from 0 to 99 (100 is exclusive)
xint = random.randint(100) 
print(xint)

# Generates a random float between 0.0 and 1.0
xfloat = random.rand() 
print(xfloat)
```

### Generating random array

```python
# Generates a 1D array of 5 random integers between 0 and 99
x = random.randint(100, size=(5)) 
print(x)

# Generates a 2D array (10 rows, 10 columns) of random floats between 0.0 and 1.0
x = random.rand(10, 10) 
print(x)
```

### Pick random no. from an array

```python
# Choice picks a single random element from the provided list
x = random.choice([3, 5, 2, 8, 1, 65]) 
print(x)

# Choice generates a 2x10 array filled with random selections from the list
x = random.choice([3, 5, 2, 8, 1, 65], size=(2, 10)) 
print(x)
```

# Data distribution

### Data distribution

```python
# Generates an array of 100 elements based on set probabilities (p)
x = random.choice([1, 3, 5, 6], p=[0.1, 0.4, 0.3, 0.2], size=(100)) 
print(x)

# Generates a 2x50 array. Notice 3 has a probability of 0.0, so it will never appear.
arr = random.choice([1, 3, 5, 2, 6], p=[0.1, 0.0, 0.5, 0.3, 0.1], size=(2, 50))
print(arr)
```

### Random permutation

```python
arr=np.array([2, 4, 6, 3, 6])
random.shuffle(arr)
#will shuffle the positions of the elemets in the array randomly everytime the code is ran
print(arr)
#random.shuffle() makes changes to the org. array

newarr = np.array([4, 3, 6, 2, 9])
print(random.permutation(newarr))
print(newarr)
#random.permutation() will shuffle the elements but will not change the original array
#hence the output of print(newarr) is always the same
```

### random.shuffle() vs random.permuatation()

* random.shuffle() changes the original array
* random.permutation() just makes a view of the arrya and shuffles the elements of that
