## Generating random integer and float number

```python
from numpy import random
xint = random.randint(100) #will pick a random integer between 1 and 100
print(xint)
xfloat = random.rand() #will pick a random float between 0 and 1
print(xfloat)
```

## Generating random array

```python
x = random.randint(100, size=(5)) #will generate an array of five random numbers between 1 and 100
print(x)
x = random.rand(10, 10) #will generate a 2d array of ten random numbers between 100 and 200
print(x)
```

## Pick random no. from an array

```python
x = random.choice([3, 5, 2, 8, 1, 65]) #The choice command will pick a random no. from the arrya and store it in the varoable x
print(x)
x = random.choice([3, 5, 2, 8, 1, 65], size=(2, 10)) #The choice command will pick a random no. from the arryay and store it in the varoable x
#The size parameter will help in generating an array of a specific no. of rows and columns consisting of random no. given i the array
print(x)
```
