# Matplotlib Tutorial

### Matplotlib plotting

```python
import matplotlib.pyplot as plt
import numpy as np

x_axis = np.array([1, 10])
y_axis = np.array([7, 20])
#x_axis contains the points that should be present on the x or the horizontal axis
#y_axis contains the points that should be present on the y or the vertical axis

plt.plot(x_axis, y_axis)
#plot() makes the graph

plt.show()
#show() displays the graph
```

### Plotting without the line (only the marks)

```python
#initializing the number of points
x_axis = np.array([1, 5])
y_axis = np.array([5, 10])

plt.plot(x_axis, y_axis, "o")
#the "o" parameter tells the program just to generate the rings or the marks not the line

plt.show()
```

### Plotting graph with multiple points

```python
#les suppose that a graph went like: (0, 0) - (4, 2) then (4, 2) - (6, 0)
x_axis = np.array([0, 4, 6])
y_axis = np.array([0, 2, 0])

plt.plot(x_axis, y_axis)
#generating the graph

plt.show()
#showing the graph

#les suppose that a graph went like: (0, 0) - (4, 30) then (4, 30) - (8, 30) then (8, 30) - (10, 0)
x_axis = np.array([0, 4, 8, 10])
y_axis = np.array([0, 30, 30, 0])

plt.plot(x_axis, y_axis)
#generating the graph

plt.show()
#showing the graph

#les suppose that a graph went like: (0, 0) - (5, 3) then (5, 3) - (7, 3) then (7, 3) - (9, 7)
x_axis = np.array([0, 5, 7, 9])
y_axis = np.array([0, 3, 3, 7])

plt.plot(x_axis, y_axis)
#generating the graph

plt.show()
#showing the graph
```

### Default x points

```python
#if we dont initialize the x points , it will be taken as [1, 2, 3, 4,...]
#les plot the 3rd graph from the above block without x points

#x_axis = np.array([1, 2, 3, 4,...])
y_axis = np.array([0, 3, 3, 7])

plt.plot(y_axis)

plt.show()
```

```python
#les plot a graph by highlighting the points
#it goes: (0,0) - (1, 5) - (3, 10) - (5, 10) - (7, 0) - (9, -10) - (12, -10) - (13, 0)

x_points = np.array([0, 1, 3, 5, 7, 9, 12, 13])
y_points = np.array([0, 5, 10, 10, 0, -10, -10, 0])

plt.plot(x_points, y_points, marker = "o")
#the marker parameter adds marker and "o" means rings or dots

plt.show()

#instead of a dot or a ring, a star can aslo be added by changing the value from "o" to "*" in markers para
#les plot the same graph

x_points = np.array([0, 1, 3, 5, 7, 9, 12, 13])
y_points = np.array([0, 5, 10, 10, 0, -10, -10, 0])

plt.plot(x_points, y_points, marker = "*")
#the marker parameter adds marker and "*" means star

plt.show()

#for reference, have a look at the marker reference table on https://www.w3schools.com/python/matplotlib_markers.asp
```

### Formatting the graphs

```python
#graphs can be formatted by giving a fmt parameters in the plt.plot() statement
#it conains <marker>|<type of the line of the graph>|<colour of the line>

#les plot another plot
#it goes - (0, 4) - (2, 4) - (2, -1) - (4, -1) - (4, 2) - (6, 2) - (6, 0)

x_points = np.array([0, 2, 2, 4, 4, 6, 6])
y_points = np.array([4, 4, -1, -1, 2, 2, 0])

#using the fmt parameter
#using dot as the markers, generating dotted line of red colour
plt.plot(x_points, y_points, "o--r")

plt.show()

x_points = np.array([0, 2, 2, 4, 4, 6, 6])
y_points = np.array([4, 4, -1, -1, 2, 2, 0])

#using dot as the markers, generating dotted line of red colour
#"ms" is the  marker size
#"mec" stands for marker colour which forms an edge of the markers of that colour
#"mfc" stands for the colour that should be filled in the marker
#in the field of colour holding paras, the supported hexadecial balue and colour names can also be used
plt.plot(x_points, y_points, marker = "o", ms = 10, mec = "r", mfc = "r")

plt.show()
#for reference, have a look at the line reference table on https://www.w3schools.com/python/matplotlib_markers.asp
#for reference, have a look at the colour reference table on https://www.w3schools.com/python/matplotlib_markers.asp
```

### Matplotlib lines

```python
#it goes: (0,0) - (1, 5) - (3, 10) - (5, 10) - (7, 0) - (9, -10) - (12, -10) - (13, 0)

x_points = np.array([0, 1, 3, 5, 7, 9, 12, 13])
y_points = np.array([0, 5, 10, 10, 0, -10, -10, 0])

#linestyle or ls can be used to chnage the type of then line
#dotted can be written as ":"
#dashed can be written as "--"
plt.plot(x_points, y_points, linestyle = "dotted")
plt.show()

plt.plot(x_points, y_points, ls = "dashed")
plt.show()

#for reference, have a look at line styles table at https://www.w3schools.com/python/matplotlib_line.asp

#the line's width can aslo be set using the linewidth para which is shortened to lw
#les plot the same graph

x_points = np.array([0, 1, 3, 5, 7, 9, 12, 13])
y_points = np.array([0, 5, 10, 10, 0, -10, -10, 0])

#linestyle or ls can be used to chnage the type of then line
#dotted can be written as ":"
#dashed can be written as "--"
plt.plot(x_points, y_points, linestyle = "dotted", linewidth = 15.2)
plt.show()

plt.plot(x_points, y_points, lw = 15.2)
plt.show()
```

### Plotting multiple lines

```python
#two sets of x and y points can be created an it can be given to plt.plot() separately to plot such graph

y1 = np.array([0, 40, 80, 120, 160])

y2 = np.array([40, 60, 80, 100, 120])

plt.plot(y1)
plt.plot(y2)

plt.show()
```

### Matplotlib titles and label

```python
#labels and titles can be added using xlabel and ylabel commands
#les take the previous example or graph

y1 = np.array([0, 40, 80, 120, 160])

y2 = np.array([40, 60, 80, 100, 120])

plt.plot(y1)
plt.plot(y2)

plt.xlabel("Time (h) ------------->")
plt.ylabel("Displacement (km) ------------->,")

plt.show()

#a title can aslo be given to the graph using title()
#les the graph above

y1 = np.array([0, 40, 80, 120, 160])

y2 = np.array([40, 60, 80, 100, 120])

plt.plot(y1)
plt.plot(y2)

plt.title("Question 6 pg. no. 55")
plt.xlabel("Time (h) ------------->")
plt.ylabel("Displacement (km) ------------->,")

plt.show()

#the title can be alligned by the loc para
#supported allignments: "left", "right" and "center"
#the default is center

y1 = np.array([0, 40, 80, 120, 160])

y2 = np.array([40, 60, 80, 100, 120])

plt.plot(y1)
plt.plot(y2)

plt.title("Question 6 pg. no. 55", loc = "left")
plt.xlabel("Time (h) ------------->")
plt.ylabel("Displacement (km) ------------->,")

plt.show()
```

### Adding grids to the graphs

```python
#grids can be added by adding plt.grid()
#les plot a graph

x_points = np.array([0, 5, 10, 15, 20, 25, 30])
y_points = np.array([10, 15, 20 , 20, 30, 15, 0])

plt.grid()

plt.plot(x_points, y_points, marker = "o")

plt.show()

#we can specify the grid that should be displayed
#suppoeted are "x", "y" and "both"
#default is "both"
#it is done by axis=" " in plt.grid()

#les plot a graph

x_points = np.array([0, 5, 10, 15, 20, 25, 30])
y_points = np.array([10, 15, 20 , 20, 30, 15, 0])

plt.grid(axis = "x")

plt.plot(x_points, y_points, marker = "o")

plt.show()

#les plot another graph

x_points = np.array([0, 5, 10, 15, 20, 25, 30])
y_points = np.array([10, 15, 20 , 20, 30, 15, 0])

plt.grid(axis = "y")

plt.plot(x_points, y_points, marker = "o")

plt.show()

#the color, ls and the lw can also be set for the grid

#les plot a graph

x_points = np.array([0, 5, 10, 15, 20, 25, 30])
y_points = np.array([10, 15, 20 , 20, 30, 15, 0])

plt.title("V - T Graph")
plt.xlabel("Time (s) ----->")
plt.ylabel("Velocity (m/s) ----->")

plt.grid(axis = "both", color = "black", ls = "--", lw = "1")

plt.plot(x_points, y_points, marker = "o")

plt.show()
```
