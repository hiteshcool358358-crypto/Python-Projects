### Performing basic arithematical operations numpy arrays

```python
import numpy as np

arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([10, 20, 30, 40, 50])
arr3 = np.array([90, 80, 70, 60, 50])

#use "+" for adding arrays
arr1 + arr2 + arr3
print(arr1 + arr2 + arr3)

#use "*" for multiplying arrays
arr1 * arr2 * arr3
print(arr1 * arr2 * arr3)

#use "**" for multiplying arrays
arr1 * 2
arr2 * 2
arr3 * 2
print(arr1 * 2)
print(arr2 * 2)
print(arr3 * 2)
```
