### What are unfuncs

- it helps in doing operations in numpy arrays element - by - element
- it is faster that python for loops

### 1. Performing basic arithematical operations numpy arrays

```python
import numpy as np

arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([10, 20, 30, 40, 50])
arr3 = np.array([90, 80, 70, 60, 50])

#use "+" for adding arrays
arr1 + arr2 + arr3
print(arr1 + arr2 + arr3)

#use "*" for multiplying arrays
arr1 * arr2 * arr3
print(arr1 * arr2 * arr3)

#use "**" for multiplying arrays
arr1 ** 2
arr2 ** 2
arr3 ** 2
print(arr1 ** 2)
print(arr2 ** 2)
print(arr3 ** 2)
```

### 2. Advanced Mathematical Ufuncs for AI
These functions process entire arrays instantly and are heavily used in AI algorithms to transform data.

```python
# Absolute value (removes negative signs)
negative_data = np.array([-1.5, -2.0, 3.5, -4.2])
print("Absolute Values:", np.abs(negative_data))

# Clamping values (np.clip forces numbers into a min/max range)
# Extremely useful for cleaning data or handling neural network outputs
raw_scores = np.array([-10, 5, 50, 120, 85])
clipped_scores = np.clip(raw_scores, 0, 100) # Anything below 0 -> 0, above 100 -> 100
print("Clipped Scores (0-100):", clipped_scores)

# Exponential and Logarithms (The math behind neural network layers)
probabilities = np.array([1, 2, 3])
print("Exponential (e^x):", np.exp(probabilities))
```

### 3. Aggregation across Axes (axis=0 vs axis=1)
Aggregation ufuncs compress arrays into summary statistics. 
- axis=0: Looks vertically down columns
- axis=1: Looks horizontally across rows

```python
# Create a 2D matrix representing 3 students and their scores in 2 subjects
# Row 0: Student A, Row 1: Student B, Row 2: Student C
# Column 0: Math, Column 1: Science
grades_matrix = np.array([
    [85, 90],
    [40, 55],
    [75, 80]
])

# Column-wise average (axis=0) -> Average for Math and Average for Science
subject_averages = np.mean(grades_matrix, axis=0)
print("Average score per subject (Math, Science):", subject_averages)

# Row-wise average (axis=1) -> Average score for each individual student
student_averages = np.mean(grades_matrix, axis=1)
print("Average score for each student (A, B, C):", student_averages)
```

### 4. Broadcasting
How NumPy handles math between arrays of completely different shapes by stretching the smaller one.

```python
# A 3x3 matrix
matrix = np.array([
    [10, 10, 10],
    [20, 20, 20],
    [30, 30, 30]
])

# A 1D array of shape (3,)
modifier = np.array([1, 2, 3])

# NumPy automatically stretches 'modifier' down to a 3x3 shape to add it to every row!
broadcasted_result = matrix + modifier
print("Broadcasted Addition Result:\n", broadcasted_result)
```
