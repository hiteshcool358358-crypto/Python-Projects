# Machine Learning and Scipy

### Mean, median and mode

```python
import numpy
from scipy import stats
import math
from matplotlib import pyplot as plt

# let a data be:
speed = [99,86,87,88,111,86,103,87,94,78,77,85,86]

print(numpy.mean(speed))
print(numpy.median(speed))
print(stats.mode(speed))
```

### Standard deviation

- It tells that how much all the values are spread out from the mean of a data.
- It is represented by the symbol sigma i.e., σ

```python
# let some datas be:
data1 = [32,111,138,28,59,77,97] 
data2 = [86,87,88,86,87,85,86]

# its standard deviation:
print(numpy.std(data1))
print(numpy.std(data2))
```

### Variance

- The square root of variance = standard deviation.
- It is represented by the symbol sigma squared i.e., σ^2

```python
data1 = [32,111,138,28,59,77,97]
print(numpy.var(data1))
print("Variance: ", numpy.var(data1))
print("Square root of the variance: ", math.sqrt(numpy.var(data1)))
print("Standard deviation: ", numpy.std(data1))

if (math.sqrt(numpy.var(data1))) == numpy.std(data1):
    print("The square root of variance is equal to the standard deviation.")
else:
    print("The square root of variance is not equal to the standard deviation")
```

### Percentile

The answer that we get for the following is 31.0 on entering 50 as the percentage which means that 50% of the data present in the array is either equal to or less than 31.0

```python
# let a data be:
data1 = [5,31,43,48,50,41,7,11,15,39,80,82,32,2,8,6,25,36,27,61,31]

print(numpy.percentile(data1, 50))
```
