# Machine Learning and Scipy

### Mean, median and mode

```python
import numpy
from scipy import stats
import math
from matplotlib import pyplot as plt

# let a data be:
speed = [99,86,87,88,111,86,103,87,94,78,77,85,86]

print(numpy.mean(speed))
print(numpy.median(speed))
print(stats.mode(speed))
```

### Standard deviation

- It tells that how much all the values are spread out from the mean of a data.
- It is represented by the symbol sigma i.e., σ

```python
# let some datas be:
data1 = [32,111,138,28,59,77,97] 
data2 = [86,87,88,86,87,85,86]

# its standard deviation:
print(numpy.std(data1))
print(numpy.std(data2))
```

### Variance

- The square root of variance = standard deviation.
- It is represented by the symbol sigma squared i.e., σ^2

```python
data1 = [32,111,138,28,59,77,97]
print(numpy.var(data1))
print("Variance: ", numpy.var(data1))
print("Square root of the variance: ", math.sqrt(numpy.var(data1)))
print("Standard deviation: ", numpy.std(data1))

if (math.sqrt(numpy.var(data1))) == numpy.std(data1):
    print("The square root of variance is equal to the standard deviation.")
else:
    print("The square root of variance is not equal to the standard deviation")
```

### Linear regression

Regression - it is used when we try to find the relation between variables or values

```python
# we need to plot scatter points first using data:
x = [5,7,8,7,2,17,2,9,4,11,12,9,6]
y = [99,86,87,88,111,86,103,87,94,78,77,85,86]

# plotting the linear regression using the stats from scipy:
slope, intercept, r, p, std_err = stats.linregress(x, y)
def myfunc(x):
    return slope * x + intercept
mymodel = list(map(myfunc, x))
plt.scatter(x, y)
plt.plot(x, mymodel)
plt.show()
```

### R for relationship

the value of r varies from -1 to 1. 0 means no relationship while -1 and 1 means 100% relationship

```python
# let us calculate the r or relation for the data:
x = numpy.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = numpy.array([99,86,87,88,111,86,103,87,94,78,77,85,86])

slope, intercept, r, p, std_err = stats.linregress(x, y)
print(r)
```

### Polynomial regression

Whenever a linear regression does not touch any data point, it may not be suitable for normal regression but may be suitable for polynomial regression

```python
# let a data be:
x = [0,1,3,5,9,12,13]
y = [0,5,10,10,-10,-10,0]

slope, intercept, r, p, std_err = stats.linregress(x, y)
def linear_regression(x):
    return slope * x + intercept
myregress = list(map(linear_regression, x))
plt.scatter(x, y)
plt.plot(x, myregress)
plt.show()
# the linear regression generated here does not touch any scatter point hence the data is more suitable for a polynimial regression

# let us make a polynomial regression for this data:
poly_regression = numpy.poly1d(numpy.polyfit(x, y, 3))
poly_line = numpy.linspace(1, 22, 100)
plt.scatter(x, y)
plt.plot(poly_line, poly_regression(poly_line))
plt.show()
# and now this touches some scatter points
```

### Polynomial regression's bad fit

```python
x = numpy.array([89,43,36,36,95,10,66,34,38,20,26,29,48,64,6,5,36,66,72,40])
y = numpy.array([21,46,3,35,67,95,53,72,58,10,26,34,90,33,38,20,56,2,47,15])

poly_regression = numpy.poly1d(numpy.polyfit(x, y, 3))
poly_line = numpy.linspace(x.min(), x.max(), 200)

plt.scatter(x, y)
plt.plot(poly_line, poly_regression(poly_line))
plt.show()

# the line of regression is not touching any of the data point
```
