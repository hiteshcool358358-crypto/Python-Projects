### Mean, median and mode

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sn

# let a data be [99,86,87,88,111,86,103,87,94,78,77,85,86] 

speed=[99,86,87,88,111,86,103,87,94,78,77,85,86]

# calculating mean:
print(np.mean(speed))

# calculating median:
print(np.median(speed))

# calculating mode:
print(stats.mode(speed))
```

### Standard deviation and Variance

- These are numbers which tell that how spread the values aur from the mean value
- Variance is the square of the standard deviation

```python
# let the data be [32,111,138,28,59,77,97]
speed=[32,111,138,28,59,77,97]

# calculating satandard deviation
print(np.std(speed))

# calculating variance
print(np.var(speed))

# proving that variance is the square of the standard deviation
if np.std(speed) == (np.var(speed)**0.5):
    print("Variance is the square of standard deviation")
else:
    print("Variance is not the square of standard deviation")
```

### Percentile

```python
# let an array be [5,31,43,48,50,41,7,11,15,39,80,82,32,2,8,6,25,36,27,61,31]
ages = [5,31,43,48,50,41,7,11,15,39,80,82,32,2,8,6,25,36,27,61,31]

print(np.percentile(ages, 65))
# the answer is 39.0 which means that 65% people are of age 39 or younger

print(np.percentile(ages, 90))
# the answer is 61.0 which means that 90% people are of age 61 or younger
```

### Data distribution and histograms

```python
# genrating a random numpy array of 250 nos. between 0.0 and 5.0 
arr1=np.random.uniform(0.0, 5.0, 250)
print(arr1)

# les plot an histogram for arr
plt.hist(arr1)
plt.plot()
plt.show()

# les do the same but with 10000000 values
# genrating a random numpy array of 10000000 nos. between 0.0 and 5.0 
arr2=np.random.uniform(0.0, 5.0, 10000000)
print(arr2)

# les plot an histogram for arr
plt.hist(arr2)
plt.plot()
plt.show()
```

### Normal data distribution

```python
# les create a normal random data
arr = np.random.normal(1.0, 5.0, 100000)

plt.hist(arr, 10)
plt.show()

sn.displot(arr, kind="kde")
```
