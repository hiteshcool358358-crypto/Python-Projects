### Mean, median and mode

```python
import numpy as np
from scipy import stats

# let a data be [99,86,87,88,111,86,103,87,94,78,77,85,86] 

speed=[99,86,87,88,111,86,103,87,94,78,77,85,86]

# calculating mean:
print(np.mean(speed))

# calculating median:
print(np.median(speed))

# calculating mode:
print(stats.mode(speed))
```

### Standard deviation and Variance

- These are numbers which tell that how spread the values aur from the mean value
- Variance is the square of the standard deviation

```python
# let the data be [32,111,138,28,59,77,97]
speed=[32,111,138,28,59,77,97]

# calculating satandard deviation
print(np.std(speed))

# calculating variance
print(np.var(speed))

# proving that variance is the square of the standard deviation
if np.std(speed) == (np.var(speed)**0.5):
    print("Variance is the square of standard deviation")
else:
    print("Variance is not the square of standard deviation")
```
