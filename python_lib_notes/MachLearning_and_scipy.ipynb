### Mean, median and mode

```python
import numpy as np
from scipy import stats

# let a data be [99,86,87,88,111,86,103,87,94,78,77,85,86] 

speed=[99,86,87,88,111,86,103,87,94,78,77,85,86]

# calculating mean:
print(np.mean(speed))

# calculating median:
print(np.median(speed))

# calculating mode:
print(stats.mode(speed))
```
