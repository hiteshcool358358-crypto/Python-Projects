### Pandas series and labels

- it is like a column
- it convertes an array of any dtype to a pandas series

```python
arr = [1, 5, 2, 7, 5, 2]

pd_series = pd.Series(arr)

print(pd_series)

#labels are used to get a data at particular place

#get the fourth data from the pd_series
print(pd_series[3])

#custom label names can be sed by the index para
labels = []
for i in range(len(arr)):
    labels.append(i + 1)
pd_series_CusLabels = pd.Series(arr, index = labels)

print(pd_series_CusLabels)

#labels can also be value objects like in a dictionary
data = {
    "BMW Sales" : 4,
    "Mercedes Sales" : 10,
    "Toyota Sales" : 7,
}

pd_series2 = pd.Series(data)

print(pd_series2)

#index para can also be used to extract series of some data also
pd_series_selective = pd.Series(pd_series2, index = ["Mercedes Sales", "Toyota Sales"])

print(pd_series_selective)
```

### Data frames

- they are pandas data sets

**Data frames vs series**

- data frames are tables while series is just columns

```python
#les create a data frame

data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}

myvar = pd.DataFrame(data)

print(myvar)

#extract data drom a *row* - use the loc para

#returns series
print(myvar.loc[1])

#use [[]] to return a data frame
print(myvar.loc[[1]])

#returns mutiple rows
print(myvar.loc[[0, 1]])

#naming indexes can be dome by feeding index names to an array and then feeding it to to the index para
df = pd.DataFrame(data, index = ["day1", "day2", "day3"])

print(df)

#locating the indexes can be done by the loc para
print(df.loc["day1"])

#use [[]] for getting multiple indexes
print(df.loc[["day1", "day3"]])
```
