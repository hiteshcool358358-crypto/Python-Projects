### Pandas series and labels

- it is like a column
- it convertes an array of any dtype to a pandas series

```python
arr = [1, 5, 2, 7, 5, 2]

pd_series = pd.Series(arr)

print(pd_series)

#labels are used to get a data at particular place

#get the fourth data from the pd_series
print(pd_series[3])

#custom label names can be sed by the index para
labels = []
for i in range(len(arr)):
    labels.append(i + 1)
pd_series_CusLabels = pd.Series(arr, index = labels)

print(pd_series_CusLabels)

#labels can also be value objects like in a dictionary
data = {
    "BMW Sales" : 4,
    "Mercedes Sales" : 10,
    "Toyota Sales" : 7,
}

pd_series2 = pd.Series(data)

print(pd_series2)

#index para can also be used to extract series of some data also
pd_series_selective = pd.Series(pd_series2, index = ["Mercedes Sales", "Toyota Sales"])

print(pd_series_selective)
```

### Data frames

- they are pandas data sets

**Data frames vs series**

- data frames are tables while series is just columns

```python
#les create a data frame

data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}

myvar = pd.DataFrame(data)

print(myvar)

#extract data drom a *row* - use the loc para

#returns series
print(myvar.loc[1])

#use [[]] to return a data frame
print(myvar.loc[[1]])

#returns mutiple rows
print(myvar.loc[[0, 1]])

#naming indexes can be dome by feeding index names to an array and then feeding it to to the index para
df = pd.DataFrame(data, index = ["day1", "day2", "day3"])

print(df)

#locating the indexes can be done by the loc para
print(df.loc["day1"])

#use [[]] for getting multiple indexes
print(df.loc[["day1", "day3"]])
```

### Cleaning data

any bad data can be deleted aur changed. bad data can be:
- empty data
- wrong data
- duplicate data, etc.

##### the data that will be used in this topic is new_data.csv

### Empty cells

```python
# we can remove rows by using the .dropna() attribute

df = pd.read_csv("new_data.csv")

#this creates a new dataframe
new_df = df.dropna()

print(new_df)

#the changes can be made in the org. df by stating inpplace True

df = pd.read_csv("new_data.csv")

#makes chnages in the org.
df.dropna(inplace = True)

print(df)
```

#### Replacing empty values

```python
#all the empty values of a column can be replaced by usng the fillna() method

df = pd.read_csv("new_data.csv")

#les do so for the "Calories" column
df.fillna({"Calories" : 130}, inplace=True)

print(df)

#we can replace an empty valur of a column by its average, most occuring and the middle value of that column
#mean() = avg
#median() = mid value
#mode() = most occuring

df1 = pd.read_csv("new_data.csv")

x = df1["Calories"].mean()
df1.fillna({"Calories" : x}, inplace=True)

print(df1.to_string())

df2 = pd.read_csv("new_data.csv")

x = df2["Calories"].median()
df2.fillna({"Calories" : x}, inplace=True)

print(df2.to_string())

df3 = pd.read_csv("new_data.csv")

x = df3["Calories"].mode()[0]
df3.fillna({"Calories" : x}, inplace=True)

print(df3.to_string())
```

#### Cleaning data of a wrong format

```python
#in our dataset i.e., new_data.csv, the type of date in row 22 and 26 is not correct
#les do it by the pd.to_datetime() method

df = pd.read_csv("new_data.csv")

df["Date"] = pd.to_datetime(df["Date"], format="mixed")

print(df.to_string())
```

#### Removing rows

```python
#we can use .dropna and use the subset method to clear all the rows with null value in a column

df = pd.read_csv("new_data.csv")

#all the rows with null values have been removed from date and calories columns
df.dropna(subset=["Date", "Calories"], inplace=True)

print(df)
```

#### Changing value of a row

```python
#the valu under duration and row 7 is wrong to lets set it to the mean value

df = pd.read_csv("new_data.csv")

x = df["Duration"].mean()
df.loc[7, "Duration"] = int(x)

print(df)

#if we have to change values in a large data sets, we can loop thought the columns
df_new = pd.read_csv("new_data.csv")
for x in df_new.index:
    if df_new.loc[x, "Duration"] > 120:
        df_new.loc[x, "Duration"] = 120

print(df_new)
```

#### Removing rows

```python
df = pd.read_csv("new_data.csv")

for x in df.index:
    if df.loc[x, "Duration"] > 120:
        df.drop(x, inplace=True)

print(df)
```

#### Removing dplicate values

```python
#we can use the .duplicated() method to print True or False for each row
#True = duplicate
#False = not a duplicate

df = pd.read_csv("new_data.csv")

#12 returns True which means that 12 is aduplicate of 11
print(df.duplicated())

#duplicates can be removed by the .drop_duplicates() method
df.drop_duplicates(inplace=True)

print(df)
```
