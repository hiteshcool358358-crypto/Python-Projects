# NumPy Random

### Generating random integer and float number

```python
import numpy as np
from numpy import random

# Generates a random integer from 0 to 99 (100 is exclusive)
xint = random.randint(100) 
print(xint)

# Generates a random float between 0.0 and 1.0
xfloat = random.rand() 
print(xfloat)
```

### Generating random array

```python
# Generates a 1D array of 5 random integers between 0 and 99
x = random.randint(100, size=(5)) 
print(x)

# Generates a 2D array (10 rows, 10 columns) of random floats between 0.0 and 1.0
x = random.rand(2, 50) 
print(x)
```

### Pick random no. from an array

```python
# Choice picks a single random element from the provided list
x = random.choice([3, 5, 2, 8, 1, 65]) 
print(x)

# Choice generates a 2x10 array filled with random selections from the list
x = random.choice([3, 5, 2, 8, 1, 65], size=(2, 10)) 
print(x)
```

# Data distribution

### Data distribution

```python
# Generates an array of 100 elements based on set probabilities (p)
x = random.choice([1, 3, 5, 6], p=[0.1, 0.4, 0.3, 0.2], size=(100)) 
print(x)

# Generates a 2x50 array. Notice 3 has a probability of 0.0, so it will never appear.
arr = random.choice([1, 3, 5, 2, 6], p=[0.1, 0.0, 0.5, 0.3, 0.1], size=(2, 50))
print(arr)
```

### Random permutation

```python
arr=np.array([2, 4, 6, 3, 6])
random.shuffle(arr)
#will shuffle the positions of the elemets in the array randomly everytime the code is ran
print(arr)
#random.shuffle() makes changes to the org. array

newarr = np.array([4, 3, 6, 2, 9])
x = random.permutation(newarr)
print(x.base)
print(random.permutation(newarr))
print(newarr)
#random.permutation() will shuffle the elements but will not change the original array
#hence the output of print(newarr) is always the same
```

### random.shuffle() vs random.permuatation()

* random.shuffle() changes the original array
* random.permutation() just makes a view of the array and shuffles the elements of that 

### Normal distribution

```python
import matplotlib
from matplotlib import pyplot as plt
import seaborn as sns

#plotting a displot using seborn

sns.displot([0, 1, 2, 3, 4, 5])

plt.show()

#to plot the graph witout the histogram use kind = "kde"

#plotting a displot using seborn without histogram

sns.displot([0, 1, 2, 3, 4, 5], kind = "kde")

plt.show()

#we can also add the three paras i.e., loc, scale and size which creating the arr
#loc is the mean or the centre of the graph
#scale expand or contracts the curve
#size is the size of the array (no. of rows ans columns)

arr = random.normal(loc = 1, scale = 2, size = (2, 3))
print(arr)

#generating a normal displot

sns.displot(random.normal(size = 1000), kind = "kde")
plt.show()
```

### Binomial distribution

```python
#can generate a graph of binary distribution like toss of a coin (100/2) and rollin a die (100/6 = ~16.67)
#it also has 3 bht different paras
#it has n (no. of trials), p (probability) size (no. of data units)

x = random.binomial(n = 10, p=1/6, size = (1000))
sns.displot(x, kind = "kde")

plt.show()
```

### Normal vs binomial data distribution

```python
data = {
    "normal" : random.normal(loc = 50, scale = 5, size = 100),
    "binomial" : random.binomial(n = 100, p = 0.5, size = 100)
}

sns.displot(data, kind = "kde")

plt.show()
```
